# CLAMEA experiment pipeline

This notebook is the single orchestration entry point. Each code stage lives in `scripts/` and is executed in the shared notebook namespace to preserve the original Colab/Marimo behavior. All saved cell outputs and execution counters are intentionally cleared.


```python
from pathlib import Path
import sys


def _find_project_root(start: Path) -> Path:
    candidates = [start.resolve(), *start.resolve().parents]
    for candidate in candidates:
        if (candidate / "src" / "clamea_refactor").is_dir() and (candidate / "scripts").is_dir():
            return candidate
        nested = candidate / "clamea"
        if (nested / "src" / "clamea_refactor").is_dir() and (nested / "scripts").is_dir():
            return nested
    raise FileNotFoundError("Cannot locate the refactored CLAMEA project root.")


PROJECT_ROOT = _find_project_root(Path.cwd())
SRC_DIR = PROJECT_ROOT / "src"
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from clamea_refactor import run_script
```
